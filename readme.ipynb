# 📊 Multi-Phase Quantitative Investment Screening Engine

A disciplined, professional-grade quantitative research dashboard built entirely in Python. This platform decouples heavy financial mathematical processing from the user interface, implementing a strict **three-phase evaluation framework** designed to protect capital, eliminate structural financial risk, filter out retail "hype bubbles," and isolate ideal algorithmic market entry windows.

---

## 🏛️ Architectural Framework: The 3-Phase Screening Strategy

To eliminate emotional trading and information overload, this engine passes asset baskets through three highly structured validation gates before generating execution verdicts:

### 🧱 Phase 1: Fundamental Health & Moat Filter

Disqualifies low-quality or highly distressed corporations regardless of how attractive their stock charts look. It processes a customized **4-Pillar 100-Point Point Allocation Matrix**:

* **Moat Profitability (25 pts):** Tracks Gross and Operational EBITDA Margins to verify pricing power.
* **Financial Runway Liquidity (25 pts):** Checks Cash-to-Debt and Current Ratios to guard against sudden bankruptcies or share dilutions.
* **Growth & Cash Flow Truth (25 pts):** Looks at YoY Revenue Growth and penalizes companies with negative Free Cash Flow (TTM) to filter out fake "accounting-only" profits.
* **Capital Efficiency & Pricing (25 pts):** Uses PEG Ratios (Price/Earnings-to-Growth) and Return on Equity (ROE) to ensure you aren't severely overpaying for momentum.

### ⏱️ Phase 2: Tactical Timing & Stability Monitor

Monitors short-term derivatives and volume structures to prevent buying at local market tops. It tracks real-time data using:

* **Relative Strength Index (RSI-14):** Evaluates asset overbought/oversold momentum boundaries.
* **Vectorized On-Balance Volume (OBV):** Measures institutional share accumulation vs. standard retail distribution.
* **Speculative Flow Index (SFI):** Analyzes the raw short-term Put/Call options volume ratio relative to underlying equity volume to measure derivative-driven Gamma Squeeze and pullback risks.
* **Risk Window Slider:** Allows users to dynamically switch between *Conservative* (downside protection) and *Aggressive* (momentum chasing) boundary thresholds.

### 🧠 Phase 3: Sentiment Ensemble Layer

Anchors professional market conviction and crowd psychology using:

* **Wall Street Macro Consensus:** Scrapes current consensus analyst recommendation arrays and measures exact geometric distance to mean price targets.
* **AI Narrative Sentiment Engine:** Leverages the **Google Gemini 2.5 Flash API** to dynamically scan, parse, and evaluate live financial news headline streams, compressing the current narrative into a standardized crowd-psychology **AI FOMO Index (0-100)**.

## ⚙️ Installation & Environment Settings

Follow these precise steps to set up and host this environment on your local Windows terminal machine or local IDE:

### 1. Clone the Repository & Navigate In

```bash
git clone https://github.com/YOUR_GITHUB_USERNAME/YOUR_REPOSITORY_NAME.git
cd fastapi-project

```

### 2. Set Up an Isolated Virtual Environment (`venv`)

Isolate your package versions to prevent clashing with global system python modules:

```powershell
# Create the environment on your terminal drive
python -m venv venv

# Activate the virtual environment path
.\venv\Scripts\Activate.ps1

```

### 3. Install Core Framework Libraries

Install all necessary data acquisition, technical analysis, and cloud model connectivity tools:

```powershell
.\venv\Scripts\python.exe -m pip install --upgrade yfinance
.\venv\Scripts\python.exe -m pip install streamlit pandas numpy pandas_ta plotly google-genai fastapi uvicorn

```

## 🚀 Execution & Deployment Commands

Because this dashboard relies on specialized background scripts, it cannot be hosted using a standard `python execution` shell. Run the dedicated server binary from your environment terminal:

```powershell
# Launch the main web application panel
.\venv\Scripts\streamlit.exe run dashboard.py

```

A browser window will automatically boot open on your default system browser hosted locally at `http://localhost:8501`.
